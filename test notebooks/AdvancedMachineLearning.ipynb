```python
import torch
import torch.nn as nn


class Generator(nn.Module):
    def __init__(
        self,
        noise_dim=128,
        caption_dim=256,
        hidden_dim=512,
        output_dim=784,
    ):
        super().__init__()

        # Three-layer conditional generator
        self.network = nn.Sequential(
            nn.Linear(noise_dim + caption_dim, hidden_dim),
            nn.ReLU(inplace=True),

            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(inplace=True),

            nn.Linear(hidden_dim, output_dim),
            nn.Tanh(),
        )

    def forward(self, noise, encoded_caption):
        """
        noise:          (batch_size, noise_dim)
        encoded_caption:(batch_size, caption_dim)
        """
        # Combine random noise and text embedding
        combined = torch.cat(
            [noise, encoded_caption], dim=1
        )

        # Generate output
        generated = self.network(combined)
        return generated


if __name__ == "__main__":
    torch.manual_seed(42)

    batch_size = 4
    noise_dim = 128
    caption_dim = 256

    generator = Generator(
        noise_dim=noise_dim,
        caption_dim=caption_dim,
        hidden_dim=512,
        output_dim=784,
    )

    # Sample Gaussian noise: mean=0, standard deviation=1
    noise = torch.randn(batch_size, noise_dim)

    # Example caption embeddings (replace with real text embeddings)
    encoded_caption = torch.randn(batch_size, caption_dim)

    # Forward pass
    generated_output = generator(noise, encoded_caption)

    print("Noise shape:", noise.shape)
    print("Caption shape:", encoded_caption.shape)
    print("Generated shape:", generated_output.shape)
```
